# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt
import numpy as np

# Years on the x-axis
years = np.arange(2008, 2015)

# Mock average prices for each product (A to E)
product_A = [400, 420, 430, 430, 370, 340, 330]
product_B = [370, 390, 400, 400, 360, 330, 320]
product_C = [0, 0, 100, 140, 180, 160, 200]
product_D = [0, 0, 120, 160, 200, 190, 170]
product_E = [0, 0, 0, 120, 160, 180, 150]

# Plot setup
plt.figure(figsize=(10, 6))

# Plot lines
plt.plot(years, product_A, label='A', color='gray')
plt.plot(years, product_B, label='B', color='gray')
plt.plot(years, product_C, label='C', color='gray')
plt.plot(years, product_D, label='D', color='gray')
plt.plot(years, product_E, label='E', color='gray')

# Highlight the recommended range
plt.axhspan(150, 200, color='lightgray', alpha=0.6)
plt.text(2008.1, 190, 'Recommended range', color='navy', fontweight='bold')

# Annotate labels
plt.text(2008, 405, 'A', color='black')
plt.text(2008, 375, 'B', color='black')
plt.text(2009.8, 110, 'C', color='black')
plt.text(2010.7, 150, 'D', color='black')
plt.text(2011.9, 170, 'E', color='black')

# Annotate average price point
plt.axhline(223, color='white', linestyle='--', linewidth=2)
plt.text(2014.1, 223, 'AVG', va='center', color='navy', fontweight='bold')

# Titles and axes
plt.title("Retail price over time", loc='left', fontsize=12)
plt.xlabel("Year")
plt.ylabel("Average price")
plt.xticks(years)
plt.yticks(np.arange(0, 501, 100))
plt.ylim(0, 500)

# Clean style
plt.grid(False)
plt.box(False)
plt.tight_layout()
plt.show()
```
